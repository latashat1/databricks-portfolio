# 01 - Ingest Raw Data: P&C Underwriting & Claims

## Source
Wisconsin Local Government Property Insurance Fund (LGPIF), via the open-access
"Loss Data Analytics" actuarial textbook (University of Wisconsin).
https://raw.githubusercontent.com/OpenActTextDev/LDA_Ed2/main/Data/WiscPropFund.csv

Real government open data - not Kaggle, no login/API key required. Direct CSV
download via browser (Free Edition blocks outbound calls from notebooks, so no
%sh curl), uploaded through Catalog Explorer.

## Shape
Panel/longitudinal data: one row per policy PER YEAR, 2006-2010. 5,639 rows
across 1,227 distinct policies (~4.6 years average per policy - consistent
with some policies entering/leaving the fund mid-period, which matches the
fund's real history).

## Verified
5,639 rows, 1,227 distinct PolicyNum, schema matches documented source columns.
