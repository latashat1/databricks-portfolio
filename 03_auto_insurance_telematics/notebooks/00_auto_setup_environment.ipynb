## What this notebook does

## Objects created
- Schema: `main.auto_insurance_telematics`
- Volume: `main.auto_insurance_telematics.raw_data` (landing zone for the claims CSV)
- Volume: `main.auto_insurance_telematics.checkpoints` (streaming checkpoints - both for Kafka and Autoloader)

```python
%sql
CREATE SCHEMA IF NOT EXISTS main.auto_insurance_telematics
COMMENT 'Auto insurance claims, policy, and live telematics driving-behavior data for claims and risk analytics';

CREATE VOLUME IF NOT EXISTS main.auto_insurance_telematics.raw_data
COMMENT 'Landing zone for raw claims CSV extract';

CREATE VOLUME IF NOT EXISTS main.auto_insurance_telematics.checkpoints
COMMENT 'Streaming checkpoint locations - Kafka telematics stream and Autoloader claims ingestion';
```
